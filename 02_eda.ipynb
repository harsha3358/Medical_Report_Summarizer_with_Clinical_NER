```python
from datasets import load_from_disk

dataset = load_from_disk("summarizer_dataset")

print("Dataset loaded:", len(dataset))
```

```python
sample = dataset[0]

print("ARTICLE:\n", sample["article"][:300])
print("\nSUMMARY:\n", sample["highlights"])
```

```python
article_lengths = [len(x["article"].split()) for x in dataset]
summary_lengths = [len(x["highlights"].split()) for x in dataset]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(article_lengths, bins=50, color='blue')
plt.title("Article Length Distribution")

plt.subplot(1,2,2)
plt.hist(summary_lengths, bins=50, color='green')
plt.title("Summary Length Distribution")

plt.show()
```

```python
import numpy as np

print("Article Length Stats:")
print("Mean:", np.mean(article_lengths))
print("Max:", np.max(article_lengths))
print("Min:", np.min(article_lengths))

print("\nSummary Length Stats:")
print("Mean:", np.mean(summary_lengths))
print("Max:", np.max(summary_lengths))
print("Min:", np.min(summary_lengths))
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("t5-small")

token_lengths = [len(tokenizer.encode(x["article"])) for x in dataset]

plt.hist(token_lengths, bins=50, color='purple')
plt.title("Token Length Distribution")
plt.show()

print("Max token length:", max(token_lengths))
```

# Observations
- Most articles are between 300–800 words.
- Summaries are much shorter (1–3 sentences).
- Token lengths vary significantly across samples.

# Conclusion
- We choose max_length = 256 for input sequences.
- We choose max_length = 64 for summary output.
- Padding and truncation are required for uniform input size.
